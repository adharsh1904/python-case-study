```python
employees = [
    {"id":101, "name":"Alice", "department":"Data", "salary":65000, "rating":4.5},
    {"id":102, "name":"Bob", "department":"AI", "salary":72000, "rating":4.8},
    {"id":103, "name":"Charlie", "department":"Web", "salary":58000, "rating":3.9},
    {"id":104, "name":"David", "department":"Data", "salary":61000, "rating":4.2},
    {"id":105, "name":"Eva", "department":"AI", "salary":75000, "rating":4.9},
    {"id":106, "name":"Frank", "department":"Cloud", "salary":68000, "rating":4.1},
    {"id":107, "name":"Grace", "department":"Web", "salary":59000, "rating":3.8},
    {"id":108, "name":"Helen", "department":"Data", "salary":70000, "rating":4.6}
]


search_id= int(input("Enter the Employee ID:"))

#answer of ques 1 total number of employees
print("Total Employees:")
print(len(employees))

#answer of ques 2 salary
salaries=[employee["salary"] for employee in employees]
average_salary=sum(salaries)/len(salaries)
highest_salary=max(salaries)
lowest_salary=min(salaries)
print("Average Salary:")
print(average_salary)
print("Highest Salary:")
print(highest_salary)
print("Lowest Salary:")
print(lowest_salary)

#answer of ques 3 names of employees whose salary is greater than ₹65,000 and whose performance rating is 4.5 or above.
print("Employees with Salary > ₹65,000 and Performance Rating >= 4.5:")
for employee in employees:
    if employee["salary"] > 65000 and employee["rating"] >= 4.5:
        print(employee["name"])


 #answer of ques 4 Create a dictionary with Employee ID as the key and Employee Name as the value.
employee_dictionary ={}
for employee in employees:
    employee_dictionary[employee["id"]]=employee["name"]
print("\nEmployees dictionary")
print(employee_dictionary)

#answer of ques 5 Create a list containing all employee names, sort it alphabetically, reverse it, and display the first three names.
names=[employee["name"]for employee in employees]
names.sort()
names.reverse()
print("First three names:")
print(names[:3])

#answer of ques 6 Store all Employee IDs in a tuple and display the maximum and minimum Employee ID.
employee_ids=tuple(employee["id"]for employee in employees)
print("Employee IDs:")
print(employee_ids)
print("Maximum Employee ID:")
print(max(employee_ids))
print("Minimum Employee ID:")
print(min(employee_ids))

#answer of ques 7 Create a set of unique departments and display the department names.
dept={employee["department"]for employee in employees}
print("Unique Departments:")
print(dept)

#answer of ques 8 Accept an Employee ID from the user and display the employee details if found; otherwise display "Employee Not Found".
employee_id=int(input("Enter Employee ID:"))
found=False
for employee in employees:
    if employee["id"]==employee_id:
        print("Employee found")
        print("ID:",employee["id"])
        print("Name:",employee["name"])
        print("Department:",employee["department"])
        print("Salary:",employee["salary"])
        print("Rating:",employee["rating"])
        found=True
        break
if not found:
    print("Employee Not Found")
```
